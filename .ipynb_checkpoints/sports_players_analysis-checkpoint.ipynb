```python
def get_time_slot(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'
```

```python
drop_columns = [
    'booking_id',
    'user_id',
    'venue_id',
    'booking_status',
    'cancellation_reason',
    'is_cancelled',
    'cancelled'
]
```

```python
y = df['cancelled']
```

```python
X.head()
```

```python
y.head()
```

```python
X.dtypes
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object', 'category']
).columns.tolist()
```

```python
print("Numerical:", numeric_features)
print("Categorical:", categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore'
    ))
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])
```

```python
print(y.value_counts())
print(y.value_counts(normalize=True) * 100)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

logistic_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(
        max_iter=1000,
        class_weight='balanced'
    ))
])
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)
```

```python
y_prob = logistic_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("Accuracy:", accuracy_score(y_test, y_pred))

print("Precision:", precision_score(
    y_test, y_pred, zero_division=0
))

print("Recall:", recall_score(
    y_test, y_pred, zero_division=0
))

print("F1 Score:", f1_score(
    y_test, y_pred, zero_division=0
))

print("ROC-AUC:", roc_auc_score(
    y_test, y_prob
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
df[['booking_date', 'booking_time', 'booking_datetime']].head()
```

```python
historical_columns = [
    'previous_bookings',
    'previous_cancellations',
    'previous_completed_bookings',
    'user_previous_cancellation_rate',
    'venue_previous_bookings',
    'venue_previous_cancellations',
    'venue_previous_cancellation_rate',
    'sport_previous_bookings',
    'sport_previous_cancellations',
    'sport_previous_cancellation_rate'
]

df = df.drop(
    columns=[
        col for col in historical_columns
        if col in df.columns
    ]
)
```

```python
df['previous_bookings'] = (
    df.groupby('user_id').cumcount()
)
```

```python
df['previous_cancellations'] = (
    df.groupby('user_id')['cancelled']
      .transform(
          lambda x: x.shift(1).fillna(0).cumsum()
      )
)
```

```python
shift(1)
```

```python
df['previous_completed_bookings'] = (
    df['previous_bookings'] -
    df['previous_cancellations']
)
```

```python
df['user_previous_cancellation_rate'] = (
    df['previous_cancellations'] /
    df['previous_bookings'].replace(0, 1)
)
```

```python
df['venue_previous_bookings'] = (
    df.groupby('venue_id').cumcount()
)
```

```python
df['venue_previous_cancellations'] = (
    df.groupby('venue_id')['cancelled']
      .transform(
          lambda x: x.shift(1).fillna(0).cumsum()
      )
)
```

```python
df['venue_previous_cancellation_rate'] = (
    df['venue_previous_cancellations'] /
    df['venue_previous_bookings'].replace(0, 1)
)
```

```python
df['sport_previous_bookings'] = (
    df.groupby('sport').cumcount()
)
```

```python
df['sport_previous_cancellations'] = (
    df.groupby('sport')['cancelled']
      .transform(
          lambda x: x.shift(1).fillna(0).cumsum()
      )
)
```

```python
df['sport_previous_cancellation_rate'] = (
    df['sport_previous_cancellations'] /
    df['sport_previous_bookings'].replace(0, 1)
)
```
